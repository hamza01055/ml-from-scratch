that scores 99% on a leaky split is worse than useless, because you will trust it.

## What this chapter covers

## Starter code

Enough to run and build on. The rest of the chapter fills in from here.

```python
from mlscratch.metrics import accuracy, precision_recall_f1, confusion_matrix

# A 95/5 imbalanced problem where accuracy lies to you.
n = 1000
y_true = np.zeros(n, dtype=int)
y_true[:50] = 1

y_lazy = np.zeros(n, dtype=int)          # predicts "negative" for everything

prec, rec, f1 = precision_recall_f1(y_true, y_lazy)
print(f"accuracy:  {accuracy(y_true, y_lazy):.3f}   <- looks excellent")
print(f"recall:    {rec:.3f}   <- catches zero positives")
print(f"f1:        {f1:.3f}")
print("\nconfusion matrix:\n", confusion_matrix(y_true, y_lazy, n_classes=2))
print("\nTODO: implement stratified k-fold and a bootstrap CI.")
```
