## Starter code

Enough to run and build on. The rest of the chapter fills in from here.

```python
def gini(y):
    """Gini impurity: probability of misclassifying a random sample
    if labelled according to the class distribution in this node."""
    if len(y) == 0:
        return 0.0
    _, counts = np.unique(y, return_counts=True)
    p = counts / len(y)
    return float(1.0 - np.sum(p ** 2))

def entropy(y):
    if len(y) == 0:
        return 0.0
    _, counts = np.unique(y, return_counts=True)
    p = counts / len(y)
    return float(-np.sum(p * np.log2(p + 1e-12)))

pure  = np.array([0, 0, 0, 0])
mixed = np.array([0, 0, 1, 1])
skew  = np.array([0, 0, 0, 1])

for name, y in [("pure", pure), ("50/50", mixed), ("75/25", skew)]:
    print(f"{name:>6}: gini {gini(y):.3f}   entropy {entropy(y):.3f}")

print("\nTODO: use these to score candidate splits, then build the tree.")
```
