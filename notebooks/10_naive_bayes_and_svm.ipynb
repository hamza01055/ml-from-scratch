## Starter code

Enough to run and build on. The rest of the chapter fills in from here.

```python
def hinge_loss(y, scores, C=1.0, w=None):
    """Hinge loss with optional L2 penalty. Labels must be -1 or +1.

    Zero loss once a point is correctly classified beyond the margin —
    that is what makes SVMs depend only on the support vectors.
    """
    margins = np.maximum(0.0, 1.0 - y * scores)
    loss = C * np.mean(margins)
    if w is not None:
        loss += 0.5 * np.sum(w ** 2)
    return float(loss)

y = np.array([1, 1, -1, -1])
confident = np.array([2.0, 3.0, -2.0, -4.0])     # all beyond the margin
marginal  = np.array([0.5, 0.2, -0.1, -0.6])     # inside the margin
wrong     = np.array([-1.0, -2.0, 1.0, 2.0])

print(f"confident: {hinge_loss(y, confident):.3f}   <- zero, nothing to learn from")
print(f"marginal:  {hinge_loss(y, marginal):.3f}")
print(f"wrong:     {hinge_loss(y, wrong):.3f}")
print("\nTODO: train a linear SVM by subgradient descent, then add an RBF kernel.")
```
