## Starter code

Enough to run and build on. The rest of the chapter fills in from here.

```python
from mlscratch.models import LinearRegression

# Starting point: generate data that a flexible model will overfit.
n = 20
X_true = np.linspace(-1, 1, n).reshape(-1, 1)
y_true = np.sin(3 * X_true).ravel() + rng.normal(0, 0.2, n)

def polynomial_features(X, degree):
    """Expand a single feature into [x, x^2, ..., x^degree]."""
    return np.hstack([X ** d for d in range(1, degree + 1)])

for degree in [1, 3, 15]:
    Xp = polynomial_features(X_true, degree)
    model = LinearRegression(method="normal").fit(Xp, y_true)
    train_mse = np.mean((model.predict(Xp) - y_true) ** 2)
    print(f"degree {degree:>2}: train MSE {train_mse:.5f}")

print("\nTrain error keeps dropping. That is exactly the problem.")
print("TODO: add a held-out set and watch test error turn around.")
```
