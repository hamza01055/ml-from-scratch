PCA finds the directions along which data varies most. It is linear algebra
rather than machine learning, but it underpins compression, visualisation, noise
reduction, and a surprising amount of preprocessing.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))

import numpy as np
import matplotlib.pyplot as plt

np.set_printoptions(precision=3, suppress=True)
rng = np.random.default_rng(42)
```

## What this chapter covers

## Starter code

Enough to run and build on. The rest of the chapter fills in from here.

```python
# Correlated 2D data — most of the variance lies along one diagonal.
mean = np.array([0.0, 0.0])
cov = np.array([[3.0, 2.0], [2.0, 2.0]])
X = rng.multivariate_normal(mean, cov, size=300)

Xc = X - X.mean(axis=0)
C = (Xc.T @ Xc) / (len(Xc) - 1)

eigenvalues, eigenvectors = np.linalg.eigh(C)
order = np.argsort(eigenvalues)[::-1]
eigenvalues, eigenvectors = eigenvalues[order], eigenvectors[:, order]

print("covariance:\n", C)
print("\neigenvalues:", eigenvalues)
print("explained variance ratio:", eigenvalues / eigenvalues.sum())
print("\nFirst component captures most of the variance.")
print("TODO: project, reconstruct, and plot the components on the scatter.")
```
