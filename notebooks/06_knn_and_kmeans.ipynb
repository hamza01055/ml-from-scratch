## Starter code

Enough to run and build on. The rest of the chapter fills in from here.

```python
def pairwise_distances(A, B):
    """Squared Euclidean distances between every row of A and every row of B.

    Uses the identity ||a - b||^2 = ||a||^2 - 2 a.b + ||b||^2 to avoid
    materialising an (n, m, d) intermediate array.
    """
    a2 = np.sum(A ** 2, axis=1).reshape(-1, 1)
    b2 = np.sum(B ** 2, axis=1).reshape(1, -1)
    return np.maximum(a2 - 2 * A @ B.T + b2, 0)

A = rng.normal(size=(4, 3))
B = rng.normal(size=(6, 3))
D = pairwise_distances(A, B)

# Verify against the naive definition.
naive = np.array([[np.sum((a - b) ** 2) for b in B] for a in A])
print("shape:", D.shape, " matches naive:", np.allclose(D, naive))

print("\nTODO: build KNNClassifier on top of this.")
```
